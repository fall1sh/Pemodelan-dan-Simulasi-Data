```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

# Parameter


```python
lambda_ = 5
n_days = 30   
np.random.seed(42)
```

# Simulasi

```python
daily_demand = np.random.poisson(lam=lambda_, size=n_days)
```

```python
print("HASIL SIMULASI PERMINTAAN HARIAN")
print("--------------------------------")
print(f"Parameter Poisson: λ = {lambda_} unit/hari")
print(f"Jumlah hari simulasi: {n_days} hari\n")
```

```python
print("Data Permintaan per Hari:")
for day, demand in enumerate(daily_demand, 1):
    print(f"Hari {day:2d}: {demand} unit")

print("\nStatistik Deskriptif:")
print(f"- Rata-rata permintaan: {np.mean(daily_demand):.2f} unit/hari")
print(f"- Standar deviasi: {np.std(daily_demand):.2f}")
print(f"- Permintaan minimum: {np.min(daily_demand)} unit")
print(f"- Permintaan maksimum: {np.max(daily_demand)} unit")
```

# VISUALISASI HASIL

```python
plt.figure(figsize=(12, 6))
```

# Histogram hasil simulasi

```python
plt.subplot(1, 2, 1)
bin_edges = np.arange(-0.5, max(daily_demand)+1.5, 1)
plt.hist(daily_demand, bins=bin_edges, color='skyblue', 
         edgecolor='black', alpha=0.7, density=True)
plt.title('Distribusi Permintaan Harian\n(Simulasi)')
plt.xlabel('Jumlah Permintaan (unit)')
plt.ylabel('Frekuensi Relatif')
plt.grid(axis='y', linestyle='--', alpha=0.7)
```

# Perbandingan dengan distribusi teoritis

```python
plt.subplot(1, 2, 2)
x = np.arange(0, max(daily_demand)+2)
plt.bar(x, poisson.pmf(x, lambda_), color='salmon', 
        alpha=0.7, label='Teoritis (λ=5)')
plt.scatter(daily_demand, np.zeros_like(daily_demand), 
           color='blue', marker='x', label='Data Simulasi')
plt.title('Distribusi Poisson Teoritis')
plt.xlabel('Jumlah Permintaan (unit)')
plt.ylabel('Probabilitas')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

# ANALISIS KESESUAIAN

```python
print("\nANALISIS KESESUAIAN DISTRIBUSI")
print("-------------------------------")
print("Uji Chi-Square Goodness of Fit (Contoh):")
observed = np.bincount(daily_demand)
expected = poisson.pmf(range(len(observed)), lambda_) * n_days

chi_square = np.sum((observed - expected)**2 / expected)
print(f"Nilai Chi-Square: {chi_square:.2f}")
print("(Nilai kecil menunjukkan kesesuaian yang baik)")
```
